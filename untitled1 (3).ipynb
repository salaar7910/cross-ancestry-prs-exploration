```python
import pandas as pd

# Let's check our environment
print("Python and Pandas are ready for cross-ancestry bioinformatics!")

# We will fetch a sample Polygenic Score (PGS) file directly from the PGS Catalog.
url = "https://ftp.ebi.ac.uk/pub/databases/spot/pgs/scores/PGS000001/ScoringFiles/PGS000001.txt.gz"

print("Downloading and inspecting sample scoring file from PGS Catalog...")
df = pd.read_csv(url, compression='gzip', sep='\t', comment='#')

# Quick inspection of the dataset structure
print("\n--- Dataset Info ---")
print(f"Total number of variants (SNPs) in this score: {len(df):,}")
print("\nFirst 5 rows of the scoring file:")
display(df.head())
```

```python
# Check for missing values in any column
print(df.isnull().sum())
```

```python
import numpy as np
import pandas as pd

# Using your actual DataFrame variable 'df' from cell [2]
np.random.seed(42)

n_samples = 5
n_variants = len(df)

# Randomly assign dosages: 0, 1, or 2
genotype_matrix = np.random.choice([0, 1, 2], size=(n_samples, n_variants), p=[0.6, 0.3, 0.1])

# Convert to a DataFrame using your variant rsIDs
df_genotypes = pd.DataFrame(genotype_matrix, columns=df['rsID'])

print(f"Genotype matrix shape: {df_genotypes.shape}")
df_genotypes.head()
```

```python
# Extract the effect weights (beta) as a 1D numpy array from your scoring file DataFrame 'df'
beta_weights = df['effect_weight'].values

# Compute the dot product between the genotype dosages (Individuals x Variants)
# and the effect weights (Variants,)
raw_prs_scores = np.dot(df_genotypes.values, beta_weights)

# Store results in a clean summary DataFrame
df_results = pd.DataFrame({
    'Sample_ID': [f"Ind_{i+1}" for i in range(n_samples)],
    'Raw_PRS': raw_prs_scores
})

print("Core PRS calculation complete!")
display(df_results)
```

```python
# Create a copy of your genotype table and introduce a few random missing values (NaN)
df_genotypes_na = df_genotypes.copy()
df_genotypes_na.iloc[0, 2] = np.nan
df_genotypes_na.iloc[3, 15] = np.nan

# Impute the missing NaNs with the mean dosage of each specific variant column
df_imputed = df_genotypes_na.fillna(df_genotypes_na.mean())

# Recalculate the PRS scores using the imputed data
prs_imputed = np.dot(df_imputed.values, beta_weights)

# Add the new scores back to your results table for comparison
df_results['PRS_Imputed'] = prs_imputed

print("Missing data handled and scores updated!")
display(df_results)
```

```python
# Let's simulate allele frequency differences between two populations (e.g., Pop A vs Pop B)
np.random.seed(42)

# Generate baseline allele frequencies for our 77 variants in Population 1
pop1_frequencies = np.random.uniform(0.05, 0.50, size=len(df))

# Simulate ancestry divergence by introducing random shifts for Population 2
# Some variants will drift higher, some lower
ancestry_drift = np.random.normal(0, 0.15, size=len(df))
pop2_frequencies = np.clip(pop1_frequencies + ancestry_drift, 0.01, 0.99)

# Create a comparison DataFrame to visualize the divergence
df_ancestry = pd.DataFrame({
    'rsID': df['rsID'],
    'Effect_Weight': df['effect_weight'],
    'Pop1_Freq': pop1_frequencies,
    'Pop2_Freq': pop2_frequencies,
    'Freq_Difference': np.abs(pop1_frequencies - pop2_frequencies)
})

print("Ancestry frequency divergence matrix generated!")
display(df_ancestry.head(10))
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Simulate a larger multi-ancestry cohort of individuals (e.g., 100 individuals)
np.random.seed(42)
n_cohort = 100
cohort_genotypes = np.random.choice([0, 1, 2], size=(n_cohort, len(df)), p=[0.5, 0.4, 0.1])

# Run PCA to reduce 77 variant dimensions down to 2 Principal Components (PC1 and PC2)
pca = PCA(n_components=2)
principal_components = pca.fit_transform(cohort_genotypes)

# Create a DataFrame for plotting
df_pca = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
df_pca['Population'] = ['Pop_A' if i < 50 else 'Pop_B' for i in range(n_cohort)]

# Plotting the ancestry divergence clusters
plt.figure(figsize=(8, 6))
for pop, color in zip(['Pop_A', 'Pop_B'], ['blue', 'orange']):
    subset = df_pca[df_pca['Population'] == pop]
    plt.scatter(subset['PC1'], subset['PC2'], label=pop, alpha=0.7, edgecolors='k')

plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.title('Ancestry Stratification via PCA')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
